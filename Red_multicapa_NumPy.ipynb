```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = np.array([
    [150,7.5,.10],[160,8.0,.15],[145,7.2,.05],[170,8.3,.20],[155,7.8,.10],
    [165,8.1,.15],[140,7.0,.05],[175,8.5,.20],[158,7.9,.10],[162,8.2,.15],
    [120,18.0,.90],[125,19.0,.95],[115,17.5,.85],[130,20.0,1.00],[118,18.5,.90],
    [123,19.5,.95],[110,17.0,.80],[128,20.5,1.00],[121,18.8,.90],[126,19.2,.95]
], dtype=float)
y = np.array([0]*10 + [1]*10).reshape(-1,1)

X_train, X_test, y_train, y_test = train_test_split(
    X,y,test_size=.2,random_state=42,stratify=y
)
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)

def sigmoid(z): return 1/(1+np.exp(-z))
def calcular_error(y_real,y_predicha):
    eps=1e-8
    return -np.mean(y_real*np.log(y_predicha+eps)+(1-y_real)*np.log(1-y_predicha+eps))

np.random.seed(42)
W=np.random.randn(3,1)*.01
b=0.0
lr=.1
for epoch in range(2000):
    A=sigmoid(X_train@W+b)
    error=calcular_error(y_train,A)
    dZ=A-y_train
    dW=X_train.T@dZ/len(X_train)
    db=np.mean(dZ)
    W-=lr*dW
    b-=lr*db
    if epoch%400==0:
        print(epoch, round(error,4))

probs=sigmoid(X_test@W+b)
preds=(probs>=.5).astype(int)
print("Accuracy:",np.mean(preds==y_test))
```
